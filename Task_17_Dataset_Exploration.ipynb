## 12. Interview Questions

### Q1. Why should a dataset be explored before modeling?

Dataset exploration is important because it helps us understand the
structure and characteristics of the data before building a machine
learning model. It helps identify data types, missing values,
outliers, distributions, and possible data quality problems.

### Q2. What does DataFrame.shape return?

DataFrame.shape returns a tuple containing the number of rows and
the number of columns in the DataFrame.

For example:

df.shape

returns:

(150, 5)

This means the dataset contains 150 rows and 5 columns.

### Q3. What information does info() provide?

The info() function provides important information about a DataFrame,
including the column names, number of non-null values, data types,
number of entries, and memory usage.
